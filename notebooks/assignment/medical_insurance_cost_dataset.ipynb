Column Chart:

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Medical_Insurance_Cost Dataset.csv")

avg_age = df.groupby("age")["bmi"].mean().sort_index(ascending=True)

avg_age.plot(kind="bar", color="purple", figsize=(8,6), alpha=0.6, legend=True)
plt.title("Average BMI per Age")
plt.ylabel("BMI")
plt.xlabel("Age")
plt.show()

#Reflection
print("The average BMI fluctuates across ages without a clear linear trend")
print("It peaks around age 27 and again in the mid-50s, while the early 30s show relatively lower BMI levels")
print("Overall, BMI varies noticeably between age groups rather than steadily increasing with age.")
```

Pie Chart:

```python
avg_age = df.groupby("region")["charges"].mean()

plt.figure(figsize=(8,8))
plt.pie(avg_age, labels=avg_age.index, autopct='%1.2f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.title("Region vs Average Charges")

plt.show()


#Reflection
print("The northwest region has the highest average charges at nearly 30%.")
print("The southeast and northeast are almost equal at around 23.5%, while the southwest is slightly lower at 23.1%.")
print("Overall, charges are fairly balanced across regions, with the northwest standing out.")
```

Bubble charts/Scatter plots:

```python
plt.scatter(
    df["age"], 
    df["children"], 
    s=df["children"]*90, 
    alpha=0.6
    )

plt.xlabel("Age")
plt.ylabel("Children")
plt.title("Age vs Children")

plt.show()


#Reflection
print("The chart shows that most individuals across different ages have only one child.")
print("A few people around their late 20s and late 30s stand out with three children, while larger family sizes are not common in the dataset.")
print("Overall, the number of children does not appear strongly tied to age.")
```

Box plots:

```python
plt.figure(figsize=(8,5))

df.boxplot(column="bmi", by="smoker")

plt.title("BMI Distribution by Smoker Status")
plt.ylabel("BMI")
plt.xlabel("Smoker Status")

plt.show()

#Reflection
print("The chart shows that smokers generally have a slightly higher median BMI compared to non-smokers.")
print("Both groups have a similar overall spread, but smokers display a wider upper range with some higher BMI values.")
print("This suggests that while smoking status does not drastically change BMI distribution, smokers tend to have more variability and higher BMI outliers.")
```
