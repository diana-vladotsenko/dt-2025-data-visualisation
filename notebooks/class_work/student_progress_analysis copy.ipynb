Who puts the more efforts?

```python
import pandas as pd

df= pd.read_csv("student_progress.csv")
df.head(20)
```

```python
import matplotlib.pyplot as plt

#take the student name and take their average hours studied
student =df.groupby("Student")["Hours_Studied"].mean()

student.plot(kind ="bar", color ="purple", figsize =(8,6), width =0.6, alpha =0.7, legend=True)
plt.title("Hours studied per student")
plt.ylabel("Hours_Studied")
plt.xlabel("Student")
plt.show()
```

Who perfoms the best?

```python
import matplotlib.pyplot as plt

max_score = df["Grade"].max()
print("Who perfoms the best?")
print(student.idxmax(), ". Her Max score:", max_score)

#take the student name and take their average grades
student =df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
student.plot(kind ="bar", color ="green", figsize =(8,6), width =0.6, alpha =0.7, legend=True)
plt.title("Performing per student based on their grades")
plt.ylabel("Grade")
plt.xlabel("Student")
plt.show()
```

pie chart variant:

```python
import matplotlib.pyplot as plt
import pandas as pd

df= pd.read_csv("student_progress.csv")
student_avg =df.groupby("Student")["Grade"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,7))
plt.pie(student_avg, labels=student_avg.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.title("Performing per student based on their grades")

plt.show()
```
