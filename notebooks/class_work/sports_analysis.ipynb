Bar chars for years of practice per athleate

```python
import matplotlib.pyplot as plt

df.plot(kind ="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of practice per athlete")
plt.ylabel("Years")
plt.show()
```

```python
df.plot(kind ="barh", x="Name", y="Age", legend=True)
plt.title("Age per athlete")
plt.ylabel("Age")
plt.xlabel("Name")
```

```python
plt.scatter(
    df["Age"], 
    df["Years_of_practice"], 
    s=df["Years_of_practice"]*90, 
    alpha=0.6
    )
plt.xlabel("Age")
plt.ylabel("Years_of_practice")
plt.title("Age vs years of prctice per athlete")
```

```python
#calculate quartiles and extremes(max,min)
df= pd.read_csv("sports.csv")

age_min = df["Age"].min()
age_max = df["Age"].max()
age_q1 = df["Age"].quantile(0.25)
age_q3 = df["Age"].quantile(0.75)

#print("Age -Min:", age_min, "Q1:",age_q1, "Q3:", age_q3, "Age Max:", age_max)
print("Years -Min:", age_min, ",Q1:",age_q1, ",Q3:", age_q3, ",Years Max:", age_max)

#years_min = df["Years_of_practice"].min()
#years_max = df["Years_of_practice"].max()
#years_q1 = df["Years_of_practice"].quantile(0.25)
#years_q3 = df["Years_of_practice"].quantile(0.75)

plt.figure(figsize=(10,5))
df[["Age", "Years_of_practice"]].boxplot()
#df[["Years_of_practice"]].boxplot()
plt.title("Box Plot for Age and Years_of_practice")
plt.show
```

```python
avg_age =df.groupby("Sport")["Age"].mean()

avg_age.plot(kind ="bar", legend=True)
plt.title("Age per sport")
plt.ylabel("Age")
plt.xlabel("Sport")
plt.show()
```
